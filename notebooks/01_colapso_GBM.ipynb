# Bloque 1 — Cuando el precio no camina, salta

Esta actividad muestra una trayectoria de Movimiento Browniano Geométrico (GBM) que permanece continua hasta un evento abrupto. El objetivo es observar la diferencia entre un movimiento **continuo** y un **salto discontinuo**.

**Guion breve para el docente:**

1. Proyecte la trayectoria inicial.
2. Pregunte: “¿Qué esperan que ocurra?”.
3. Ejecute la animación hasta el punto rojo.
4. Deténgase y pregunte: “¿Qué pasó aquí?”.
5. Conecte el evento con Merton (1976): cuando el mercado puede saltar, un modelo de difusión continua deja de ser suficiente.

Referencia: Merton, R. C. (1976). *Option pricing when underlying stock returns are discontinuous*. Journal of Financial Economics, 3(1-2), 125-144.

```python
# Dependencias mínimas para Google Colab
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

np.random.seed(2024)

# Simulación: GBM suave hasta el día 150; allí se agrega un salto de -12%
S0 = 100
mu = 0.05
sigma = 0.18
T = 1.0
steps = 252
dt = T / steps
jump_day = 150
jump_size = -0.12

Z = np.random.normal(0, 1, steps)
gbm_path = np.zeros(steps + 1)
gbm_path[0] = S0
for t in range(1, steps + 1):
    gbm_path[t] = gbm_path[t-1] * np.exp((mu - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * Z[t-1])

# La parte después del salto conserva la trayectoria pero aplica el desplazamiento multiplicativo
jumped_path = gbm_path.copy()
jumped_path[jump_day:] *= np.exp(jump_size)

fig, ax = plt.subplots(figsize=(10, 4.5))
line, = ax.plot([], [], color='black', lw=2)
ax.axhline(S0, color='gray', ls='--', alpha=0.5)
ax.set_xlim(0, steps)
ax.set_ylim(jumped_path.min() * 0.98, gbm_path.max() * 1.02)
ax.set_xlabel('Días')
ax.set_ylabel('Precio')
ax.set_title('GBM con un evento discontinuo')
ax.grid(alpha=0.3)
annotation = ax.annotate('', xy=(0, 0), xytext=(0, 0), arrowprops=dict(arrowstyle='->', color='black'))
annotation.set_visible(False)

def init():
    line.set_data([], [])
    annotation.set_visible(False)
    return line, annotation

def update(frame):
    x = np.arange(frame + 1)
    line.set_data(x, jumped_path[:frame + 1])
    if frame >= jump_day:
        annotation.xy = (jump_day, jumped_path[jump_day])
        annotation.xytext = (jump_day + 8, jumped_path[jump_day] * 1.05)
        annotation.set_text('¿Qué pasó aquí?')
        annotation.set_visible(True)
    return line, annotation

ani = FuncAnimation(fig, update, frames=steps + 1, init_func=init, interval=18, blit=True)
plt.close(fig)
display(HTML(ani.to_jshtml()))
```
