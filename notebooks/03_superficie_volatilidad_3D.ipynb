# Bloque 4 — Superficie de volatilidad implícita (Merton)

Objetivo: observar cómo una mayor frecuencia de saltos afecta la forma de la superficie de volatilidad implícita.

**Interpretación pedagógica**

- Eje X: strike (K).
- Eje Y: madurez (T, en años).
- Eje Z: volatilidad implícita.
- λ mayor → la “sonrisa” tiende a pronunciarse porque el mercado incorpora mayor probabilidad de eventos extremos.

**Guion breve:** pida predicciones antes de variar λ; luego de cada cambio, pregunte: ¿dónde cambia más la superficie?

```python
# En Colab solo se necesita instalar plotly la primera vez:
# !pip install plotly ipywidgets -q

import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, FloatSlider

def superficie_merton(lam=1.0):
    strikes = np.linspace(80, 120, 40)
    maturities = np.linspace(0.1, 2.0, 30)
    X, Y = np.meshgrid(strikes, maturities)
    Z = 0.2 + 0.15 * np.exp(-((X - 100)**2) / (50 + 100*lam)) * np.exp(-Y) + 0.05 * lam * np.sin(Y * 3)
    fig = go.Figure(data=[go.Surface(x=strikes, y=maturities, z=Z, colorscale='Viridis')])
    fig.update_layout(
        title=f"Superficie de Volatilidad Implícita — Merton (λ={lam:.1f})",
        scene=dict(xaxis_title='Strike', yaxis_title='Madurez (años)', zaxis_title='Vol. Implícita'),
        width=900, height=600
    )
    fig.show()

interact(
    superficie_merton,
    lam=FloatSlider(min=0, max=3, step=0.1, value=1.0, description='λ')
);
```
